```python
df = df.drop(columns=['applicant_id'])
```

```python
numeric_like_cols = [
    'age',
    'dependents',
    'work_experience_years',
    'intended_duration'
]

df[numeric_like_cols].head()
```

```python
for col in numeric_like_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
# Numerical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())

# Categorical columns
cat_cols = df.select_dtypes(include=['object']).columns
df[cat_cols] = df[cat_cols].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df[(df['age'] < 0) | (df['age'] > 100)]
```

```python
valid_age_mask = (df['age'] >= 0) & (df['age'] <= 100)
median_age = df.loc[valid_age_mask, 'age'].median()

df.loc[~valid_age_mask, 'age'] = median_age
```

```python
df[(df['age'] < 0) | (df['age'] > 100)]
```

```python
def fill_duration(row):
    if row['purpose_of_travel'] == 'Tourism':
        return 30
    elif row['purpose_of_travel'] == 'Business':
        return 15
    elif row['purpose_of_travel'] == 'Study':
        return 365
    elif row['purpose_of_travel'] == 'Work':
        return 730
    elif row['purpose_of_travel'] == 'Medical':
        return 60
    else:
        return 30

df['intended_duration'] = df.apply(fill_duration, axis=1)
```

```python
df.to_csv("../data/processed/visa_cleaned.csv", index=False)
```
